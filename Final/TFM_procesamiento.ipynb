Dentro del proceso de preparación y limpieza de los datos, resulta imprescindible **examinar las variables categóricas para identificar los distintos valores que pueden adoptar y calcular la frecuencia de cada uno de ellos**. Este procedimiento permite conocer con mayor precisión la composición del dataset, detectar categorías poco comunes y localizar posibles inconsistencias derivadas de errores en la introducción de datos.  

Para esta tarea se empleará la función `tabla_valres_unicos`, que ofrece un resumen de los valores únicos junto con el número de veces que aparecen en cada variable categórica.


```python
def tabla_valores_unicos(df):
    """
    Devuelve un DataFrame con la frecuencia absoluta y relativa de los valores únicos 
    en cada columna de `df`.

    Los valores nulos se muestran como "Nulo". El resultado tiene un MultiIndex 
    (columna, valor) y columnas:
        - "n": frecuencia absoluta
        - "%": frecuencia relativa (0 a 1)

    Args:
        df (pandas.DataFrame): DataFrame de entrada.

    Returns:
        pandas.DataFrame: Tabla de frecuencias por columna y valor único.
    """
    df_frec = pd.DataFrame()

    for col in df.columns:
        # Generare, solo para la visualización, que estos elementos se llamen "Nulo"
        series_fillna = df[col].fillna("Nulo")

        # Realizamos el conteo de cada valor y su frecuencia
        series_cont = series_fillna.value_counts()
        series_frec = series_cont/series_cont.sum()

        # Usaremos un DataFrame con MultiIndex para poder visualizar y acceder mejor a cada columna
        arrays_multiindex = [[col]* len(series_cont), series_cont.index.tolist()]
        multi_index = pd.MultiIndex.from_arrays(arrays_multiindex)

        result = pd.DataFrame({"n":series_cont.values, "%": series_frec.values}, index=multi_index)
        df_frec = pd.concat([df_frec, result], axis=0)
    
    return df_frec
```

## Tratamiento de los valores nulos
Uno de los aspectos clave de este trabajo es que se ha trabajado con **datos procedentes de la vida real**, en lugar de utilizar un *dataset* predefinido y preparado específicamente para fines académicos o de investigación. Esta decisión aporta un valor añadido, ya que acerca el análisis a un escenario más realista y representativo de los problemas que pueden encontrarse en entornos profesionales.

Sin embargo, esta elección también implica una serie de desafíos adicionales. Los datos reales suelen presentar un mayor número de **valores faltantes** e **inconsistencias**, lo que incrementa la complejidad del proceso de depuración y limpieza. 

En este contexto, el **tratamiento de los valores nulos** adquiere un papel fundamental, ya que garantiza la calidad del conjunto de datos y permite que los modelos predictivos trabajen sobre una base más sólida y fiable. De esta manera, se logra minimizar el impacto de la incompletitud de la información y se sientan las bases para obtener resultados más consistentes y precisos en las fases posteriores del estudio.

Es importante recordar que cada anuncio procede de un archivo en formato **JSON**, donde se almacenan sus diferentes características. Sin embargo, no todas estas variables se encuentran presentes en cada registro. En muchos casos, cuando una característica no contiene información relevante, se omite deliberadamente con el fin de **optimizar el uso del almacenamiento** y evitar ocupar espacio innecesario con valores vacíos.  

Esta situación incrementa la heterogeneidad del conjunto de datos y refuerza la necesidad de aplicar un tratamiento riguroso de valores nulos e inconsistencias, garantizando así que la información disponible pueda ser utilizada de manera coherente y eficaz en las etapas posteriores del análisis.

### Variables catégoricas nulas

| Variable | Nº de Nulos | Tipo | Estrategia de Imputación |
| :--- | :--- | :--- | :--- |
| **floor** | 7207 | object | Se estandarizan los valores no numéricos (ej. 'bj', 'st') a sus equivalentes numéricos (0, -1).<br> Los nulos se tratan como planta baja (0) y se crea una variable `floor_cat` para capturar el efecto no lineal de estas categorías especiales. |
| **detailedType.subTypology** | 17080 | object | Se imputa con una nueva categoría, 'Vacío', dada que no existe subtipolgia en ese caso. |
| **district** | 3450 | object | Se imputan los nulos mediante geolocalización inversa, utilizando las coordenadas para asignar el distrito correspondiente. |
| **highlight.groupDescription** | 14883 | object | Se imputa con la categoría 'Vacio', asumiendo que la ausencia de valor implica la ausencia de una promoción. |
| **neighborhood** | 10291 | object | La columna se elimina debido a su alta cardinalidad y nulos, y se reemplaza por una feature binaria, `has_neighborhood`. |
| **status** | 597 | object | Se imputan los nulos como 'New development', basándose en la correlación observada con la variable `newDevelopment`, lo que resuelve una inconsistencia en los datos. |

```python
# floor
def floor_function(value):
    """ 
    Procesa la información del campo 'floor' de un inmueble y clasifica su valor.

    Args:
        value (pandas.Series or dict): Registro que debe contener:
            - "floor": valor de planta (string).
            - "detailedType.typology": tipología del inmueble.

    Returns:
        pandas.Series: Serie con dos elementos:
            - float: planta en formato numérico (0.0–15).
            - str: categoría asignada ("Numerica" o etiqueta original).
    """
    numeric_cat = "Numerica"
    x = value["floor"]

    len_x = len(x)

    numeric = [c for c in x if c.isdigit()]
    alnum = [c for c in x if not c.isalnum()]

    if len(numeric) == len_x:
        res_num = float(x)
        res_cat = numeric_cat

    elif alnum:
        if alnum[0] in [".", ","]:
            res_num = float(x)
        elif alnum[0] == "+":
            res_num = float("".join(numeric))
        elif alnum[0] == "-":
            res_num = float(x)
        res_cat = numeric_cat

    else:
        res_num = 0.0
        if value["detailedType.typology"] in ["chalet", "countryHouse"]:
            res_cat = numeric_cat
        else:
            res_cat = x

    if res_num > 15:
        res_num = 15
    return pd.Series([res_num, res_cat])
```

```python
# floor
df["floor"] = df["floor"].fillna("Nulo")
df_floor = df.apply(floor_function, axis=1)
df["floor"], df["floor_cat"] = df_floor[0], df_floor[1]

# detailedType.subTypology
df["detailedType.subTypology"] = df["detailedType.subTypology"].fillna("Vacío")

# highlight.groupDescription
df["highlight.groupDescription"] = df["highlight.groupDescription"].fillna("Vacío")

# status
df["status"] = df["status"].fillna("newdevelopment")

# Neighborhood
df["has_neighborhood"] = (~df["neighborhood"].isna()).astype(int)
```

# Diferenciacion de variables
Con el fin de facilitar el manejo de la información, se procederá a **dividir las variables en función de su tipología**. Esta clasificación permite organizar el conjunto de datos de una manera más estructurada, lo que a su vez simplifica las tareas de preprocesamiento, análisis y modelado.

De este modo, cada tipo de variable podrá ser tratado de manera específica y adecuada a sus características particulares. Por ejemplo, las variables numéricas requerirán transformaciones distintas a las aplicadas sobre variables categóricas.

En consecuencia, esta organización previa favorece un trabajo más ágil y sistemático en las etapas posteriores del proyecto, reduciendo la complejidad y evitando posibles errores en el tratamiento de los datos.


```python
from itertools import zip_longest
# variables booleanas
variable_bool = info_types[info_types["Valores únicos"]<3].index.to_list()

# variables numericas
variables_num = df.select_dtypes(include=["number"]).columns.tolist()

# variables categoricas
categoricas = []
for col in df.columns:
    if df[col].dtypes == "object":
        if df[col].nunique(dropna=False) <20:
            categoricas.append(col)
df[categoricas] = df[categoricas].astype("category")
variables_cat = df.select_dtypes(include=["category"]).columns.tolist()


print(f"{"variable_bool":50}{"variables_num":<50}{"variables_cat":<10}")
print("------------------------------------------------------------------------------------------------------------------------")
for variable in zip_longest(variable_bool, variables_num, variables_cat, fillvalue=""): 
    print(f"{variable[0]:50}{variable[1]:<45}{variable[2]:<10}")
```

# Tratamiento de *outliers*

La presencia de *outliers* puede distorsionar de manera significativa los resultados del análisis y afectar negativamente el rendimiento de los modelos predictivos. Por este motivo, es fundamental aplicar un procedimiento sistemático para su detección y corrección.

En este trabajo se ha definido la función `tratar_outliers(df, columnas_a_tratar, features_para_imputar)`, cuyo objetivo es:

1. **Detectar** los valores atípicos en las columnas especificadas.  
2. **Reemplazar** dichos valores por `NaN`, con el fin de eliminar su influencia directa en el conjunto de datos.  
3. **Imputar** posteriormente los valores faltantes mediante un modelo de **Random Forest**, lo que permite estimar los valores de manera robusta y preservar la coherencia del conjunto de datos.  

El resultado final es un nuevo *DataFrame* depurado, libre de valores extremos que pudieran sesgar el análisis o comprometer la capacidad predictiva de los modelos.


```python

def tratar_outliers(df, columnas_a_tratar, features_para_imputar):
    """
    Detecta outliers, los reemplaza por NaN y los imputa usando RandomForest,
    devolviendo un nuevo DataFrame limpio.
    """
    df_procesado = df.copy()

    # Detección y Reemplazo de Outliers
    for col in columnas_a_tratar:
        if abs(df_procesado[col].skew()) < 1:
            criterio1 = abs((df_procesado[col] - df_procesado[col].mean()) / df_procesado[col].std()) > 3
        else:
            mad = sm.robust.mad(df_procesado[col].dropna())
            criterio1 = abs((df_procesado[col] - df_procesado[col].median()) / mad) > 8

        q1, q3 = df_procesado[col].quantile([0.25, 0.75])
        criterio2 = ~df_procesado[col].between(q1 - 3 * (q3 - q1), q3 + 3 * (q3 - q1))
        df_procesado.loc[criterio1 & criterio2, col] = np.nan

    # Imputación de Nulos con RandomForest
    for col in columnas_a_tratar:
        if df_procesado[col].isnull().sum() == 0:
            continue

        # Preparar los datos para el modelo de imputación
        train = df_procesado[df_procesado[col].notnull()]
        test = df_procesado[df_procesado[col].isnull()]

        X_train = train[features_para_imputar].fillna(train[features_para_imputar].mean())
        y_train = train[col]
        X_test = test[features_para_imputar].fillna(train[features_para_imputar].mean())

        # Entrenar y predecir
        rf = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
        rf.fit(X_train, y_train)
        predictions = rf.predict(X_test)

        # Asignar las predicciones de vuelta al DataFrame
        df_procesado.loc[df_procesado[col].isnull(), col] = predictions

    return df_procesado
```

```python

features_para_imputar = [
    'size', 'floor', 'exterior', 'hasLift',
    'parkingSpace.hasParkingSpace', 'parkingSpace.isParkingSpaceIncludedInPrice',
    'parkingSpace.parkingSpacePrice', 'newDevelopment', 'newDevelopmentFinished',
    'has_neighborhood', 'hasVideo', 'hasPlan', 'has3DTour',
    'has360', 'hasStaging', 'topPlus', 'showAddress'
]
df = tratar_outliers(
    df,
    columnas_a_tratar=['price', 'priceByArea', 'size'],
    features_para_imputar=features_para_imputar
)
```
