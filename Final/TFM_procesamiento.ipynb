Otro aspecto fundamental dentro del proceso de limpieza de datos consiste en **identificar los posibles valores que puede adoptar cada variable categórica y realizar un conteo de su frecuencia de aparición**. Para ello usaremos la función de `tabla_valres_unicos`

```python
def tabla_valores_unicos(df):
    """
    Devuelve un DataFrame con la frecuencia absoluta y relativa de los valores únicos 
    en cada columna de `df`.

    Los valores nulos se muestran como "Nulo". El resultado tiene un MultiIndex 
    (columna, valor) y columnas:
        - "n": frecuencia absoluta
        - "%": frecuencia relativa (0 a 1)

    Args:
        df (pandas.DataFrame): DataFrame de entrada.

    Returns:
        pandas.DataFrame: Tabla de frecuencias por columna y valor único.
    """
    df_frec = pd.DataFrame()

    for col in df.columns:
        # Generare, solo para la visualización, que estos elementos se llamen "Nulo"
        series_fillna = df[col].fillna("Nulo")

        # Realizamos el conteo de cada valor y su frecuencia
        series_cont = series_fillna.value_counts()
        series_frec = series_cont/series_cont.sum()

        # Usaremos un DataFrame con MultiIndex para poder visualizar y acceder mejor a cada columna
        arrays_multiindex = [[col]* len(series_cont), series_cont.index.tolist()]
        multi_index = pd.MultiIndex.from_arrays(arrays_multiindex)

        result = pd.DataFrame({"n":series_cont.values, "%": series_frec.values}, index=multi_index)
        df_frec = pd.concat([df_frec, result], axis=0)
    
    return df_frec
```

```python
# floor
def floor_function(value):
    """ 
    Procesa la información del campo 'floor' de un inmueble y clasifica su valor.

    Args:
        value (pandas.Series or dict): Registro que debe contener:
            - "floor": valor de planta (string).
            - "detailedType.typology": tipología del inmueble.

    Returns:
        pandas.Series: Serie con dos elementos:
            - float: planta en formato numérico (0.0–15).
            - str: categoría asignada ("Numerica" o etiqueta original).
    """
    numeric_cat = "Numerica"
    x = value["floor"]

    len_x = len(x)

    numeric = [c for c in x if c.isdigit()]
    alnum = [c for c in x if not c.isalnum()]

    if len(numeric) == len_x:
        res_num = float(x)
        res_cat = numeric_cat

    elif alnum:
        if alnum[0] in [".", ","]:
            res_num = float(x)
        elif alnum[0] == "+":
            res_num = float("".join(numeric))
        elif alnum[0] == "-":
            res_num = float(x)
        res_cat = numeric_cat

    else:
        res_num = 0.0
        if value["detailedType.typology"] in ["chalet", "countryHouse"]:
            res_cat = numeric_cat
        else:
            res_cat = x

    if res_num > 15:
        res_num = 15
    return pd.Series([res_num, res_cat])
```

```python
# floor
df["floor"] = df["floor"].fillna("Nulo")
df_floor = df.apply(floor_function, axis=1)
df["floor"], df["floor_cat"] = df_floor[0], df_floor[1]

# detailedType.subTypology
df["detailedType.subTypology"] = df["detailedType.subTypology"].fillna("Vacío")

# highlight.groupDescription
df["highlight.groupDescription"] = df["highlight.groupDescription"].fillna("Vacío")

# status
df["status"] = df["status"].fillna("newdevelopment")

# Neighborhood
df["has_neighborhood"] = (~df["neighborhood"].isna()).astype(int)
```

### Clustering de las variables geograficas

Un aspecto relevante en el análisis de la vivienda es el **barrio/ zona geografíca** en el que se ubica cada inmueble, pues determina o está correlacionado con múltiples factores importantes para el estudio: 
* Seguridad del barrio
* Transporte y movilidad
* Proximidad a puntos de interés (hospitales, centros educativos, etc.)
* Clase socioeconómica del barrio
* Distancia al centro de Madrid

Los datos recopilados originalmente están segmentados en un número excesivo de zonas geográficas, por lo que su uso directo como una única variable resulta poco fiable y poco informativo para los modelos. Para resolver esta fragmentación y obtener una variable más manejable y representativa, se optó por agrupar las zonas mediante técnicas de *clustering*.

En el *Jupyter Notebook* `TFM_Clustering` se describen con detalle los pasos metodológicos seguidos y la justificación para la elección de `k = 50`. En este documento solo se incluye la sección de código correspondiente a la fase de conclusión del proceso de agrupamiento.


```python
# Realización del Clustering basado en centroides (KMeans)

# Coordenadas de referencia (Puerta del Sol)
lat_centro, lon_centro = 40.416775, -3.703790

# Distancia euclidea a la Puerta del Sol
df["distancia_al_centro"] = np.sqrt((df["latitude"]-lat_centro)**2 + 
                                    (df["longitude"]-lon_centro)**2)

# Seleccionar y escalar variables para el clustering geográfico
geo_features = ["latitude", "longitude", "distancia_al_centro"]
X_geo = df[geo_features]

# Escalamos las variables 
scaler = StandardScaler()
X_geo_scaled = scaler.fit_transform(X_geo)

# Realizamos el KMeans
# Usamos un k_value de 50, gracias a la investigación previa 
k_value = 50
kmeans = KMeans(n_clusters=k_value, random_state=42, n_init="auto")
df["Zona_cluster"] = kmeans.fit_predict(X_geo_scaled)
```

# Tratamiento de Outliers

```python

def tratar_outliers(df, columnas_a_tratar, features_para_imputar):
    """
    Detecta outliers, los reemplaza por NaN y los imputa usando RandomForest,
    devolviendo un nuevo DataFrame limpio.
    """
    df_procesado = df.copy()

    # Detección y Reemplazo de Outliers
    for col in columnas_a_tratar:
        if abs(df_procesado[col].skew()) < 1:
            criterio1 = abs((df_procesado[col] - df_procesado[col].mean()) / df_procesado[col].std()) > 3
        else:
            mad = sm.robust.mad(df_procesado[col].dropna())
            criterio1 = abs((df_procesado[col] - df_procesado[col].median()) / mad) > 8

        q1, q3 = df_procesado[col].quantile([0.25, 0.75])
        criterio2 = ~df_procesado[col].between(q1 - 3 * (q3 - q1), q3 + 3 * (q3 - q1))
        df_procesado.loc[criterio1 & criterio2, col] = np.nan

    # Imputación de Nulos con RandomForest
    for col in columnas_a_tratar:
        if df_procesado[col].isnull().sum() == 0:
            continue

        # Preparar los datos para el modelo de imputación
        train = df_procesado[df_procesado[col].notnull()]
        test = df_procesado[df_procesado[col].isnull()]

        X_train = train[features_para_imputar].fillna(train[features_para_imputar].mean())
        y_train = train[col]
        X_test = test[features_para_imputar].fillna(train[features_para_imputar].mean())

        # Entrenar y predecir
        rf = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
        rf.fit(X_train, y_train)
        predictions = rf.predict(X_test)

        # Asignar las predicciones de vuelta al DataFrame
        df_procesado.loc[df_procesado[col].isnull(), col] = predictions

    return df_procesado
```

```python

features_para_imputar = [
    'size', 'latitude', 'longitude', 'floor', 'exterior', 'hasLift',
    'parkingSpace.hasParkingSpace', 'parkingSpace.isParkingSpaceIncludedInPrice',
    'parkingSpace.parkingSpacePrice', 'newDevelopment', 'newDevelopmentFinished',
    'has_Discount', 'has_neighborhood', 'hasVideo', 'hasPlan', 'has3DTour',
    'has360', 'hasStaging', 'topPlus', 'showAddress'
]
df = tratar_outliers(
    df,
    columnas_a_tratar=['price', 'priceByArea', 'size'],
    features_para_imputar=features_para_imputar
)
```
