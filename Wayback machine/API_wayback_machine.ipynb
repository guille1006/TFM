```python
import requests

def check_wayback_availability(url):
    endpoint = "https://archive.org/wayback/available"
    params = {"url": url}

    try:
        response = requests.get(endpoint, params=params)
        response.raise_for_status()
        data = response.json()

        # Verifica si hay una versión archivada
        snapshot = data.get("archived_snapshots", {}).get("closest", None)
        if snapshot and snapshot.get("available", False):
            archived_url = snapshot["url"]
            timestamp = snapshot["timestamp"]
            print(f"✅ Versión archivada encontrada:")
            print(f"📅 Fecha: {timestamp}")
            print(f"🔗 URL archivada: {archived_url}")
        else:
            print("❌ No se encontró ninguna versión archivada.")

    except requests.RequestException as e:
        print(f"❌ Error al consultar la API: {e}")

# ▶️ Prueba con un sitio web
check_wayback_availability("https://example.com")
```
